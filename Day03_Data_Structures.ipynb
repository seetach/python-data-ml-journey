Lists,Tuples,Dictionaries and Sets are called as PYTHON DATA STRUCTURES

```python
#List
#It contains multiple data
Students = ["seeta","chandhu","maha","sekhar"]
#Accessing the element
print(Students[0])
print(Students[0:3])
print(Students[1:3])
print(Students[:3])
print(Students[1:])

#list operations
#Add Item(.Append)
Students.append("mattaparti")
print(Students)

#Remove Item (.remove)
Students.remove("maha")
print(Students)

#Loop through List
for Student in Students :
  print(Student)

```

```python
#Example
files = [
    "sales.csv",
    "customers.csv",
    "products.csv",

]

#Append
files.append("orders.csv")
print(files)

#Remove
files.remove("products.csv")
print(files)

for file in files :
  print(file)
```
