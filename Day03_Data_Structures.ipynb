Lists,Tuples,Dictionaries and Sets are called as PYTHON DATA STRUCTURES

```python
#List
#It contains multiple data
Students = ["seeta","chandhu","maha","sekhar"]
#Accessing the element
print(Students[0])
print(Students[0:3])
print(Students[1:3])
print(Students[:3])
print(Students[1:])

#list operations
#Add Item(.Append)
Students.append("mattaparti")
print(Students)

#Remove Item (.remove)
Students.remove("maha")
print(Students)

#Loop through List
for Student in Students :
  print(Student)

```

```python
#Example
files = [
    "sales.csv",
    "customers.csv",
    "products.csv",

]

#Append
files.append("orders.csv")
print(files)

#Remove
files.remove("products.csv")
print(files)

for file in files :
  print(file)
```

```python
#Tuples () data can't be modified
#List will use [] data can be modified
students = ("seeta","chandu","maha","sekhar")
print(students)
print(students[1])
print(students[0:2])

for student in students :
  print(student)
```

```python
#Sets --- It always stores unique values only , Duplicate values will be removed
nums = {10,20,30}
print(nums)

nums1 = {10,20,30,10,20}
print(nums1)

customer_ids = {101,102,105,101,103,105}
unique_customers = set(customer_ids)
print(unique_customers)
```

```python
#Dictionaries ---- It stores data in the form of KEY : VALUE
student = {
    "name" : "seeta",
    "age" :  "25",
    "department" : "Data engineering"
}

print(student["name"])
```

```python
employee = {
    "id" : 101,
    "name" : "seeta",
    "department" : "data engineering"
}

print(employee["id"])
```
