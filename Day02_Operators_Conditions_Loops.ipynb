##Operators
1)Arthmetic Operators

```python
a=4
b=2

print(a + b) #add
print(a - b) #sub
print(a * b) #mul
print(a / b) #div
print(a % b) #mod
print(a ** b) #power
```

```python
#input version
num1 = 10
num2 = 5
print("Addition:", num1 + num2)
print("subtraction:",num1 - num2)
print("multiplication:",num1 * num2)
print("division:",num1 / num2)
print("modulus:",num1 % num2)
print("power:",num1 ** num2)
```

```python
#comparsion operators
a=10
b=20

print(a == b)
print(a !=b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

```python
##If statement
age = 20

if age >= 18:
    print("Eligible to vote")
```

```python
##If else statement
a = 16

if a >= 18:
    print("TRUE")
else:
    print("false")
```

#Loop
A loop is used when you want to repeat a task multiple times.


```python
#for loop
for i in range(5):
    print(i)
```

```python
# range(start,end) , always starts counting from "0"
for i in range(1,6):
    print(i)
```

```python
#While Loop
#It repeats until the condition is false
count = 1
while count < 5:
    print(count)
    count +=1
```

```python
num = 6
for i in range(1,6):
 print(num, "x", i, "=", num * i)
```
